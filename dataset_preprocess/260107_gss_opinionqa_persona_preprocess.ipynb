### GSS Preprocess

```python
import pandas as pd
import csv
import json
```

```python
data_dir='./dataset/gss/gss_dataset.csv'
question_dir='./dataset/gss/gss_useful_multiple_choice_questions.csv'
desc_dir='./dataset/gss/gss_user_attributes.csv'
data=pd.read_csv(data_dir)
question=pd.read_csv(question_dir)
```

```python
desc=[]
desc_dict={}
desc_val={}
attr_list=[]
with open(desc_dir) as file:
    csv_file=csv.reader(file)
    for lines in csv_file:
        desc.append(lines)

for d in desc[1:]:
    desc_dict[d[0]]=d[1]
    desc_val[d[0]]=d[2]
    attr_list.append(d[0].lower())
```

```python
questions=[]
with open(question_dir) as file:
    csvfile=csv.reader(file)
    for lines in csvfile:
        questions.append(lines)
```

```python
question_dict={}
q_list=[]
q_num_dict={}
q_pos_dict={}
for d in questions[1:]:
    question_dict[d[0].lower()]=d[1]
    q_list.append(d[0].lower())
    q_num_dict[d[0].lower()]=d[2]
    q_pos_dict[d[0].lower()]=d[3]
```

```python
data=[]
with open(data_dir) as file:
    csvfile=csv.reader(file)
    for lines in csvfile:
        data.append(lines)
```

```python
data[0].index('grass')
```

```python
output_file=[]
for idx, d in enumerate(data[1:]):
    out={}
    out['id']=idx
    out['feature']={}
    out['responses']=[]
    for i in range(len(data[0])):
        if data[0][i].lower() in attr_list:
            out['feature'][data[0][i]]=d[i]
        elif data[0][i].lower() in q_list:
            response={}
            response['Question_var']=data[0][i].lower()
            response['Question']=question_dict[data[0][i].lower()]
            response['Number_of_Answers']=q_num_dict[data[0][i].lower()]
            response['Possible_Answers']=q_pos_dict[data[0][i].lower()]
            response['Response']=d[i]
            out['responses'].append(response)
    output_file.append(out)
    
```

```python
output_file[0]
```

```python
with open('./dataset/GSS/GSS_per_user_with_gt.json', 'w') as f:
    json.dump(output_file, f)
```

```python
save=[]
for d in output_file:
    save.append(d['responses'])
```

```python
with open('./dataset/GSS/GSS_per_user.json', 'w') as f:
    json.dump(save, f)
```

### OpinionQA preprocess

```python
wave='27' #26, 32, 34, 36, 41, 42, 43,45, 49, 50, 54, 82, 92 29      27 not included in Suh paper
```

```python
info_dir=f'./dataset/opinionqa/American_Trends_Panel_W{wave}/info.csv'
meta_dir=f'./dataset/opinionqa/American_Trends_Panel_W{wave}/metadata.csv'
response_dir=f'./dataset/opinionqa/American_Trends_Panel_W{wave}/responses.csv'
```

### Attribute Indexing

```python
desc=[]
desc_dict={}
desc_val={}
attr_list=[]
with open(meta_dir) as file:
    csv_file=csv.reader(file)
    for lines in csv_file:
        desc.append(lines)

for d in desc[1:]:
    attr_list.append(d[1].lower())
    desc_val[d[1].lower()]=(d[2])
    
```

### Question Indexing 

```python
questions=[]
with open(info_dir) as file:
    csvfile=csv.reader(file)
    for lines in csvfile:
        questions.append(lines)
```

```python
question_dict={}
q_list=[]
q_num_dict={}
q_pos_dict={}
for d in questions[1:]:
    question_dict[d[3].lower()]=d[5]
    q_list.append(d[3].lower())
    q_pos_dict[d[3].lower()]=d[6]
```

### Response Indexing

```python
data=[]
with open(response_dir) as file:
    csvfile=csv.reader(file)
    for lines in csvfile:
        data.append(lines)
```

```python
output_file=[]
for idx, d in enumerate(data[1:]):
    out={}
    out['wave']=wave
    out['id']=idx
    out['feature']={}
    out['responses']=[]
    for i in range(len(data[0])):
        if data[0][i].lower() in attr_list:
            out['feature'][data[0][i]]=d[i]
        elif data[0][i].lower() in q_list:
            response={}
            response['Question_var']=data[0][i].lower()
            response['Question']=question_dict[data[0][i].lower()]
            response['Possible_Answers']=q_pos_dict[data[0][i].lower()]
            response['Response']=d[i]
            out['responses'].append(response)
    output_file.append(out)
    
```

```python
with open(f'./dataset/opinionqa/American_Trends_Panel_W{wave}/opinionqa_per_user_with_gt.json', 'w')  as f:
    json.dump(output_file, f)
```

```python
save=[]
for d in output_file:
    save.append(d['responses'])
```

```python
with open(f'./dataset/opinionqa/American_Trends_Panel_W{wave}/opinionqa_per_user.json', 'w')  as f:
    json.dump(save, f)
```
