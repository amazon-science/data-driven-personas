```python
import json
import copy
```

### OP 26 Demo Matching

```python
persona_source="opinionqa_W26"
form="newfree"
persona_dir=f"./src/out/ours_persona/persona/claude3.7/{persona_source}_personas_opinionqafeatures_allreviews_{form}.json"
ref_dir=f"./src/out/ours_persona/persona/claude3.7/{persona_source}_personas_opinionqafeatures_allreviews_{form}_gt.json"
all_persona_dir=f"./src/out/ours_persona/persona/claude3.7/{persona_source}_personas_allreviews_{form}.json"
source_dir=f"./dataset/opinionqa/grouped/{persona_source}_per_user_with_gt.json"
save_dir=f"./src/out/ours_persona/persona/claude3.7/{persona_source}_personas_opinionqafeatures_allreviews_{form}_gtmatch.json"
```

```python
with open(all_persona_dir) as f:
    all_persona=json.load(f)
```

```python
with open(persona_dir) as f:
   persona=json.load(f) 
```

```python
with open(ref_dir) as f:
    ref_persona=json.load(f)
```

```python
assert(len(ref_persona)==len(persona))
```

```python
for per, ref in zip(persona, ref_persona):
    new_generated_personas=[]
    if len(ref['generated_persona']):
        for gen_persona in ref['generated_persona']:
            user_idx=(gen_persona['user_idx'])
            save_persona=copy.deepcopy(gen_persona)
            bio=all_persona[user_idx]['short_bio']
            save_persona['short_bio']=bio
            new_generated_personas.append(save_persona)
    per['generated_persona']=new_generated_personas
```

```python
for per, ref in zip(persona, ref_persona):
    assert(len(per['generated_persona'])==len(ref['generated_persona']))
```

```python
with open(save_dir, 'w') as f:
    json.dump(persona, f)
```
